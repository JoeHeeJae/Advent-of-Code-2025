```python
# raw 문자열을 줄 단위로 split하고, 빈 줄 제거
lines = [line for line in raw.strip().split("\n") if line.strip()]

# 문자 단위 리스트로 변환 → 2D grid
grid = [list(line) for line in lines]

R = len(grid)
C = len(grid[0])

print("행:", R, "  열:", C)
```

### Part 1

```python
def count_at_3x3_centered(r, c):
    cnt = 0
    for dr in (-1, 0, 1):
        for dc in (-1, 0, 1):
            nr = r + dr
            nc = c + dc
            if 0 <= nr < R and 0 <= nc < C:
                if grid[nr][nc] == '@':
                    cnt += 1
    return cnt

accessible = 0

for r in range(R):
    for c in range(C):
        if grid[r][c] != '@':
            continue

        total_at = count_at_3x3_centered(r, c)

        # 🔥 기준 변경: @가 5개 이상이면 blocked / 4개 이하이면 accessible
        if total_at <= 4:
            accessible += 1

print("접근 가능한 롤 개수 =", accessible)
```

### Part 2

```python
def count_at_3x3_centered(r, c):
    cnt = 0
    for dr in (-1, 0, 1):
        for dc in (-1, 0, 1):
            nr = r + dr
            nc = c + dc
            if 0 <= nr < R and 0 <= nc < C:
                if grid[nr][nc] == '@':
                    cnt += 1
    return cnt


total_removed = 0

while True:
    to_remove = []

    # 이번 턴에서 제거할 롤 찾기
    for r in range(R):
        for c in range(C):
            if grid[r][c] != '@':
                continue

            total_at = count_at_3x3_centered(r, c)

            # 🔥 접근 가능 기준: 3×3 내 @ ≤ 4 → 제거 가능
            if total_at <= 4:
                to_remove.append((r, c))

    # 더 제거할 게 없으면 종료
    if not to_remove:
        break

    # 제거 수행
    for r, c in to_remove:
        grid[r][c] = '.'
        total_removed += 1

print("총 제거된 롤 개수 =", total_removed)
```
