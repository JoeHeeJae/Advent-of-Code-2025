### Part 1

```python
# 범위 파싱
ranges = [tuple(map(int, r.split("-"))) for r in ranges.split(",")]

def is_double_repeat(n):
    s = str(n)
    if len(s) % 2 != 0:  # 짝수 자리수만 허용
        return False
    mid = len(s) // 2
    return s[:mid] == s[mid:]  # 앞 절반 == 뒤 절반

invalid_numbers = []

for start, end in ranges:
    for num in range(start, end + 1):
        if is_double_repeat(num):
            invalid_numbers.append(num)

print("잘못된 숫자 목록:", invalid_numbers)
print("개수:", len(invalid_numbers))
print("합계:", sum(invalid_numbers))
```

### Part 2

```python
# 범위 파싱
ranges = [tuple(map(int, r.split("-"))) for r in ranges.split(",")]

def is_repeated_sequence(n):
    s = str(n)
    L = len(s)

    # 최소 2번 이상 반복되어야 함
    # 가능한 블록 길이를 모두 시도 (1 ~ L//2)
    for k in range(1, L // 2 + 1):
        if L % k != 0: 
            continue  # 길이가 딱 나눠떨어져야 함
        block = s[:k]
        if block * (L // k) == s:
            return True
    return False

invalid_numbers = []

for start, end in ranges:
    for num in range(start, end + 1):
        if is_repeated_sequence(num):
            invalid_numbers.append(num)

print("잘못된 숫자 개수:", len(invalid_numbers))
print("합계:", sum(invalid_numbers))
```
